## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

````python
#Common method to show chart of different kinds

def plot_chart(df, chart_type, col1=None, col2=None, kde=False, figsize=(6,4)):

    plt.figure(figsize=figsize)

    if chart_type == "pie":
        df[col1].value_counts().plot.pie(autopct='%1.1f%%')
        plt.ylabel("")
        plt.title(f"{col1} Distribution")

    elif chart_type == "bar":
        sns.countplot(data=df, x=col1)
        plt.title(f"{col1} Count")

    elif chart_type == "hist":
        sns.histplot(df[col1], kde=kde)
        plt.title(f"{col1} Distribution")

    elif chart_type == "box":
        sns.boxplot(x=df[col1])
        plt.title(f"{col1} Box Plot")

    elif chart_type == "scatter":
        sns.scatterplot(data=df, x=col1, y=col2)
        plt.title(f"{col1} vs {col2}")

    elif chart_type == "heatmap":
        sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
        plt.title("Correlation Heatmap")

    else:
        print("Unsupported chart type")

    plt.tight_layout()
    plt.show()
````
